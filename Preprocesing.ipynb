# Preprocesamiento de datos Vambe Clients

```python
df = pd.read_csv('vambe_clients.csv')
```

```python
print(f"Forma del dataframe: {df.shape}")
df.head()
```

```python
df.describe()
```

```python
len(df["Nombre"].unique())
```

```python
print("Valores nulos por columna:")
print(df.isnull().sum())
print(f"\nTotal de valores nulos: {df.isnull().sum().sum()}")
```

```python
df = df[['Vendedor asignado', 'closed', 'Transcripcion']]
print(f"\nNueva forma del dataframe: {df.shape}")
print("\nPrimeras filas:")
df.head()
```

Porcentaje de closed por vendedor

```python
vendedor_stats = df.groupby('Vendedor asignado').agg({
    'closed': ['count', 'sum']
}).round(2)

vendedor_stats.columns = ['Total', 'Cerrados']
vendedor_stats['Porcentaje_Cerrados'] = (vendedor_stats['Cerrados'] / vendedor_stats['Total'] * 100).round(2)
vendedor_stats = vendedor_stats.sort_values('Porcentaje_Cerrados', ascending=False)

print("Porcentaje de clientes cerrados por vendedor:")
print(vendedor_stats)
```
